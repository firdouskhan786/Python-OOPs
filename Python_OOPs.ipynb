#Q1. What is Object-Oriented Programming (OOP)?


Ans. Object-oriented programming (OOP) is a programming paradigm that organizes code into objects, which are instances of classes. It bundles data (attributes) and the functions (methods) that operate on that data together, which helps create more reusable, flexible, and organized code. This approach is widely used in software development for building applications by modeling real-world entities as objects.  

The core principles and concepts of OOP are:

Objects and Classes:

Class: A blueprint or template for creating objects. It defines the common structure and behavior shared by all objects of that type [1, 2].
Object: An instance of a class. It is a real-world entity with state (data) and behavior (methods) [1].

Core Pillars of OOP:

Encapsulation: The mechanism of binding the data (variables) and the code (methods) that operate on the data into a single unit (an object). It is used to hide the internal representation, often limiting direct access to some components [1, 2].

Inheritance: The mechanism by which a new class (subclass or derived class) can inherit properties and behaviors (fields and methods) from an existing class (superclass or base class). This promotes code reusability [1, 2].

Polymorphism: The ability of an object to take on many forms. In OOP, it allows methods to perform different functions depending on the object or data type being used, often through method overriding or overloading [1, 2].

Abstraction: The concept of hiding the complex implementation details and showing only the essential features of an object to the user. It helps in managing complexity by presenting a simplified view [1, 2].

#Q2 What is a class in OOP?

Ans. A class in Object-Oriented Programming (OOP) serves as a blueprint or template for creating objects. It defines the structure, common properties (attributes or fields), and behaviors (methods or functions) that all objects of a particular type will possess.

Here are the key characteristics and purposes of a class:

Defines Structure: A class specifies what kind of data an object can hold. For example, a Car class might define attributes like color, make, model, and year.

Defines Behavior: It also defines the actions an object can perform. The Car class might include methods like start_engine(), accelerate(), and brake().

Encapsulates Data and Methods: It bundles the data and the methods that operate on that data into a single, self-contained unit.

Instantiation: The process of creating a concrete object from a class is called instantiation. An object is an instance of a class. You could create multiple objects from the same Car class, such as car1 (red, Toyota, Camry) and car2 (blue, Honda, Accord), each with its own unique data but sharing the same blueprint.

Promotes Reusability: Classes allow developers to define a type once and create many instances of it, which is fundamental to code reusability and organization in OOP.





#Q3 What is an object in OOP?


Ans. An object in Object-Oriented Programming (OOP) is a fundamental, self-contained unit that represents a real-world entity or concept. It is a concrete instance of a class.

An object bundles together:

State (Attributes/Fields): The data associated with the object. This is what the object knows about itself. For example, in a Car object, the state might be color: "red", make: "Toyota", and speed: 60 mph.

Behavior (Methods/Functions): The actions the object can perform or the functions that operate on its data. This is what the object does. For the Car object, behaviors might include start_engine(), accelerate(), or brake().

In simple terms:

A class is the blueprint or template (e.g., the architectural drawing for a house).

An object is the actual, physical house built from that blueprint.
Multiple objects can be created from the same class, each having its own unique state but sharing the same defined behaviors and structure.

### Q1. Create a parent class Animal with a method speak() that prints a generic message. Create a child class Dog that overrides the speak() method to print "Bark!".

### Q2. Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectangle from it and implement the area() method in both.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

# Example usage
circle = Circle(5)
print("Circle Area:", circle.area())

rectangle = Rectangle(4, 6)
print("Rectangle Area:", rectangle.area())
```

### Q3. Implement a multi-level inheritance scenario where a class Vehicle has an attribute type. Derive a class Car and further derive a class ElectricCar that adds a battery attribute.

```python
class Vehicle:
    def __init__(self, type):
        self.type = type

class Car(Vehicle):
    def __init__(self, type, brand):
        super().__init__(type)
        self.brand = brand

class ElectricCar(Car):
    def __init__(self, type, brand, battery):
        super().__init__(type, brand)
        self.battery = battery

# Example usage
obj = ElectricCar("Car", "Tesla", "100 kWh")
print(f"Type: {obj.type}, Brand: {obj.brand}, Battery: {obj.battery}")
```

### Q5. Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes balance and methods to deposit, withdraw, and check balance.

```python
class BankAccount:
    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient funds!")

    def check_balance(self):
        print("Balance:", self.__balance)

# Example usage
account = BankAccount()
account.deposit(1000)
account.withdraw(500)
account.check_balance()
```

### Q7. Create a class MathOperations with a class method add_numbers() to add two numbers and a static method subtract_numbers() to subtract two numbers.

```python
# Code
class MathOperations:
    @classmethod
    def add_numbers(cls, a, b):
        return a + b

    @staticmethod
    def subtract_numbers(a, b):
        return a - b

# Example usage
print("Addition:", MathOperations.add_numbers(5, 3))
print("Subtraction:", MathOperations.subtract_numbers(5, 3))
```

### Q10. Demonstrate operator overloading by creating a class Vector and overriding the __add__ method to add two vectors.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

# Example usage
v1 = Vector(2, 3)
v2 = Vector(4, 5)
result = v1 + v2
print(result)
```

### Q12. Implement a class Student with attributes name and grades. Create a method average_grade() to compute the average of the grades.

```python
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def average_grade(self):
        return sum(self.grades) / len(self.grades)

# Example usage
student = Student("Alice", [90, 80, 85])
print(f"Average Grade: {student.average_grade()}")
```

### Q14. Create a class Employee with a method calculate_salary() that computes the salary based on hours worked and hourly rate. Create a derived class Manager that adds a bonus to the salary.

```python
class Employee:
    def __init__(self, hours, hourly_rate):
        self.hours = hours
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours * self.hourly_rate

class Manager(Employee):
    def __init__(self, hours, hourly_rate, bonus):
        super().__init__(hours, hourly_rate)
        self.bonus = bonus

    def calculate_salary(self):
        return super().calculate_salary() + self.bonus

# Example usage
manager = Manager(160, 30, 500)
print(f"Salary: {manager.calculate_salary()}")
```

### Q16. Create a class Animal with an abstract method sound(). Create two derived classes Cow and Sheep that implement the sound() method.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Cow(Animal):
    def sound(self):
        print("Moo!")

class Sheep(Animal):
    def sound(self):
        print("Baa!")

# Example usage
cow = Cow()
sheep = Sheep()
cow.sound()
sheep.sound()
```

### Q18. Create a class House with attributes address and price. Create a derived class Mansion that adds an attribute number_of_rooms.

```python
class House:
    def __init__(self, address, price):
        self.address = address
        self.price = price

class Mansion(House):
    def __init__(self, address, price, number_of_rooms):
        super().__init__(address, price)
        self.number_of_rooms = number_of_rooms

# Example usage
mansion = Mansion("123 Luxury St", 5000000, 10)
print(f"Address: {mansion.address}, Price: {mansion.price}, Rooms: {mansion.number_of_rooms}")
```
